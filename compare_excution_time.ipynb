```python
with open('numbers.txt','w') as f:
  for _ in range(100000):
    f.write(str(_)+'\n')
```

```python
# Method 1: Simple For Loop

import numpy as np
import time

start_time = time.time()
sum_of_squares = 0
with open('numbers.txt', 'r') as f:
  for line in f:
    number = int(line.strip())
    if number % 2 == 0:
      sum_of_squares += number**2
end_time = time.time()
print("Method 1 (For Loop): Sum of squares =", sum_of_squares)
print("Execution time:", end_time - start_time, "seconds")
```

```python
# Method 2: NumPy Vectorized Operation 
start_time = time.time()
numbers = np.loadtxt('numbers.txt', dtype=int)  # Load once
even_numbers = numbers[numbers % 2 == 0]
sum_of_squares = np.sum(even_numbers ** 2)
end_time = time.time()
print("\nMethod 2 (NumPy): Sum of squares =", sum_of_squares)
print("Execution time:", end_time - start_time, "seconds")
```

```python
# Method 3: List Comprehension
start_time = time.time()
with open('numbers.txt', 'r') as f:
  numbers = [int(line.strip()) for line in f]
sum_of_squares = sum(number**2 for number in numbers if number % 2 == 0)
end_time = time.time()
print("\nMethod 3 (List Comprehension): Sum of squares =", sum_of_squares)
print("Execution time:", end_time - start_time, "seconds")
```
